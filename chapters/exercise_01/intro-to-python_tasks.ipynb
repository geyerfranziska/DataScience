```raw
*
* *
* * *
* * * *
* * * * *
* * * *
* * *
* *
*
```

````python
for i in range(5):
  zeile = ""
  for j in range(i+1):
    zeile = zeile + "* "
  print(zeile)
for i in range(4):
  zeile = ""
  for j in range(4-i):
    zeile = zeile + "* "
  print(zeile)

for i in range(1,6):
  zeile = ""
  for j in range(i):
    zeile = zeile + "* "
  print(zeile)
for i in range(4,0, -1):
  zeile = ""
  for j in range(i):
    zeile = zeile + "* "
  print(zeile)
````

### Write a Python function that takes an int as a parameter and checks if the number is prime or not.

````python
def PrimeTester(zahl):
  if zahl < 2:
    return False

  for i in range(2, zahl):
    if zahl % i == 0:
      return False
  return True

PrimeTester(34757)
````

### Use your function to create a list that contains all primes from 2 to 100

````python
PrimeListe = [i for i in range(2, 101) if PrimeTester(i)]
PrimeListe
````

### Write a function that simulates dice rolls (6-sided dices)
- Write a function that has the number of dice rolls as the input parameter
- The output is a map with the counts how often each result of a roll (1-6) occurred
- You can use the randomint() function from the random package

````python
# load the random package
import random
````

````python
def Würfle(anzWürfe):
  würfelStat = {i: 0 for i in range(1,7)}
  for i in range(anzWürfe):
    zufallszahl = random.randint(1,6)
    würfelStat[zufallszahl] = würfelStat[zufallszahl] + 1
  return würfelStat

Würfle(1000)
````

### Combine two dictionaries by adding values for common keys.
d1 = {'a': 100, 'b': 200, 'c':300}

d2 = {'a': 300, 'b': 200, 'd':400}

Sample output: {'a': 400, 'b': 400, 'd': 400, 'c': 300}

````python
def addDict(dic1, dic2):
  combiDic = {}
  print( list(dic1.keys()) + list(dic2.keys()))
  for k in list(dic1.keys()) + list(dic2.keys()):
    if k not in combiDic:
      combiDic[k] = dic1.get(k, 0) + dic2.get(k, 0)
  return combiDic


d1 = {'a': 100, 'b': 200, 'c':300}
d2 = {'a': 300, 'b': 200, 'd':400}
addDict(d1, d2)
````
